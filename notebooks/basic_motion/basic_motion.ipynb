### Importing the Robot class

To get started programming JetBot, we'll need to import the ``Robot`` class.  This class

```python
from jetbot import Robot
import time
```

Now that we've imported the ``Robot`` class we can initialize the class *instance* as follows. 

```python
robot = Robot()
```

## Initial Tests
It is important to go through these initial tests, where we ensure that our robot class is working as expected. Run them all, one by one and check that the robot is showcasing the correct corresponding movement.

The robot's front is camera, that is where it looks from. To make sure the movement is correct, take the camera as a reference. For example, a left turn should make the left wheel spin forwards while the right wheel spins backwards.

**Test Failure** If any of the test fails and the robot isn't moving as it should, stop now and ask for help! Each robot is slightly different, there may be something that needs changing in the code!

### Test 1: Turn Left
You should see the robot spin towards the left.

```python
robot.left(speed=0.7)
time.sleep(0.5)
robot.stop()
```

### Test 1.1: Turn Left (individual control)
This test should also make the robot turn left. It is important to test both movement options to ensure the rest of the software will work appropriately.

```python
robot.set_motors(0.7,-0.7)
time.sleep(0.5)
robot.stop()
```

### Test 2: Turn Right
You should see the robot spin towards the right.

```python
robot.right(speed=0.7)
time.sleep(0.5)
robot.stop()
```

### Test 3: Forward
You should see the robot move forwards slightly.

```python
robot.forward(0.7)
time.sleep(0.2)
robot.stop()
```

### Test 3.1: Forward (individual control)
This test should also make the robot move forward slightly. It is important to test both movement options to ensure the rest of the software will work appropriately.

```python
robot.set_motors(0.7, 0.7)
time.sleep(0.2)
robot.stop()
```

### Test 4: Backward
You should see the robot move backwards slightly.

```python
robot.backward(0.7)
time.sleep(0.2)
robot.stop()
```
